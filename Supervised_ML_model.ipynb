##Project Context       

This notebook builds on my project, AI-Driven Digital Job Cards System, which explores how digital workflows and predictive maintenance could support clinical engieering operations.

For this exercise I am using the public AI4I 2020 Predictive Maintenance Dataset as a proxy dataset.

##Problem Definition

**Problem type**: Binary Classification

**Target**: Machine failure, where 0= no recorded failure and 1=recorded failure.

**Objective**-Train supervised machine learning models to predict the machine failure label from selected operating measurements and compare their performance on unseen test data.

```python
import pandas as pd
from google.colab import files

uploaded=files.upload()

df=pd.read_csv("ai4i2020.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nTarget values:")
print(df["Machine failure"].value_counts())
```

```python
features=["Air temperature [K]","Process temperature [K]","Rotational speed [rpm]","Torque [Nm]","Tool wear [min]"]

x=df[features]

y=df["Machine failure"]

print("Feature shape:", x.shape)
print("Target shape:", y.shape)
print("\nFirst five feature rows:")
print(x.head())
print("\nFirst five target values:")
print(y.head())

```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=42)

print("Training features:", x_train.shape)
print("Testing features:", x_test.shape)
print("\nTraining target distribution:")
print(y_train.value_counts())
print("\nTesting target distribution:")
print(y_test.value_counts())

```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logistic_model=make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000,random_state=42))

logistic_model.fit(x_train, y_train)

print("Logistic Regression model trained successfully")
```

```python
y_pred=logistic_model.predict(x_test)

print("First 20 actual outcomes:")
print(y_test.head(20).tolist())

print("\nFirst 20 predicted outcomes:")
print(y_pred[:20])
```

```python
from sklearn.metrics import(confusion_matrix,classification_report,balanced_accuracy_score)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nBalanced Accuracy Score:", balanced_accuracy_score(y_test, y_pred))

print("Balanced Accuracy:",round(balanced_accuracy_score(y_test,y_pred),4))
```

```python

from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(
    max_depth=5,
    class_weight="balanced",
    random_state=42
)

tree_model.fit(x_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
from sklearn.metrics import(confusion_matrix,classification_report,balanced_accuracy_score)
tree_pred=tree_model.predict(x_test)
print("Decision Tree Confusion Matrix:")
print(confusion_matrix(y_test,tree_pred))
print("\nDecision Tree Classification Report:")
print(classification_report(y_test,tree_pred,zero_division=0))
print("\nDecision Tree Balanced Accuracy:",round(balanced_accuracy_score(y_test,tree_pred),4))
```


## Comparison of Model Performance

| Evaluation Metric | Logistic Regression | Decision Tree |
|:------------------|:------------------:|:-------------:|
| Accuracy | 97.3% | 91.8% |
| Precision (Class 1: Failure) | 64% | 25% |
| Recall (Class 1: Failure) | 26.2% | 83.6% |
| F1-Score (Class 1: Failure) | 0.37 | 0.38 |
| Balanced Accuracy | 62.88% | 87.81% |


## Model Comparison and Discussion

I trained and evaluated Logistic Regression and Decision Tree classifiers using the AI4I 2020 Predictive Maintenance Dataset. Both models were evaluated on the same test set of 2,000 records.

Logistic Regression achieved an accuracy of 97.3% and a failure recall of 26.2%. Although its accuracy was high, it detected only 16 of the 61 actual failures. This demonstrates why accuracy alone is not sufficient when working with imbalanced datasets.

The Decision Tree achieved an accuracy of 91.8%, a failure recall of 83.6%, and a balanced accuracy of 87.81%. It correctly detected 51 of the 61 failures, but it also produced 155 false positives. Its failure precision was 25%, meaning that only one in four of its failure predictions was correct.

Overall, the Decision Tree was better at detecting failures, while Logistic Regression produced fewer false alarms. The Decision Tree achieved a slightly higher failure F1-score (0.38 compared with 0.37), despite the difference in their precision and recall.

These results demonstrate the importance of selecting evaluation metrics according to the problem being solved. For predictive maintenance, recall is important because missed failures may have serious consequences, but precision also matters because excessive false alarms can lead to unnecessary inspections.

The dataset is industrial rather than hospital-based, so these findings demonstrate a machine-learning workflow and do not establish real-world performance on medical equipment.
