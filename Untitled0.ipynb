**AI-Driven Digital Job Cards System**

1.**Problem definition.**

Medical equipment maintenance in large hospitals can be delayed by the use of physical job cards. After a repair is completed, the job card may need to be physically taken to the relevant department head for approval, which can delay the return of equipment to service. Maintenance records stored in paper files also make it difficult to quickly identify recurring equipment failures.

The proposed project is an AI-driven digital job card system that aims to digitize the maintenance workflow, improve access to equipment maintenance history, and support proactive maintenance.

**Intended AI/ML outcome**
The machine learning component of the project will use equipment and maintenance-related data to identify patterns associated with equipment failure and predict the likelihood of future failure. This information can support earlier maintenance decisions and help reduce equipment downtime.

2.**Data requierments.**

To support the machine learning component of the proposed system, the following data would ideally be collected from digital job cards and equipment records:

- Equipment information: equipment type, asset ID, age, and usage information.
- Maintenance history: previous maintenance dates, number of previous breakdowns, fault descriptions, repair frequency, and parts replaced.
- Failure information: whether the equipment experienced a failure after previous maintenance.
- Repair information: repair duration and maintenance cost.
- Workflow information: repair completion time and approval/sign-off time.

These data would allow the system to learn patterns associated with equipment failure and support proactive maintenance decisions.

For this assignment, a public predictive-maintenance dataset will be used because real hospital equipment maintenance records are not available for this project.
